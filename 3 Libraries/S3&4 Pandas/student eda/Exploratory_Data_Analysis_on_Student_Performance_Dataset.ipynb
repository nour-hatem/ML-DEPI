## 2️⃣ Data Cleaning

### Tasks:
- Ensure text-based columns follow a consistent format
- Remove or ignore columns that do not add analytical value
- Standardize categorical values if needed

```python
# cat_cols = df.select_dtypes(include=['object']).columns
cat_cols_problms = ['gender', 'country', 'residence', 'prevEducation']

for col in cat_cols_problms:
  print(f'unique for {col}:', df[col].unique())

for col in ['gender', 'country', 'residence', 'prevEducation']:
    df[col] = df[col].str.lower().str.strip()

df['gender'] = df['gender'].replace({'m': 'male', 'f': 'female'})
df['country'] = df['country'].apply(lambda x: 'south africa' if x == 'rsa' else x)
df['prevEducation'] = df['prevEducation'].replace({'highschool':'high school','barrrchelors': 'bachelors', 'diplomaaa': 'diploma'})
df['residence'] = df['residence'].replace({
    'bi-residence': 'bi residence',
    'biresidence': 'bi residence',
    'bi_residence': 'bi residence'
})

print('='*50)
print('After cleaning:')
for col in cat_cols_problms:
  print(f'unique for {col}', df[col].unique())
```

```python
df = df.drop(columns=['fNAME', 'lNAME'])
df.columns
```
