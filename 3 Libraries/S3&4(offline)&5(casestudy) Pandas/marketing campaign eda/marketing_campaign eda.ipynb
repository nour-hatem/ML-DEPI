```python
df.drop(['ID', 'Z_CostContact', 'Z_Revenue'], axis=1, inplace=True) # ID doesnt matter, Z_CostContact and Z_Revenue contains only 1 number f them

df.describe().round(2)
```

```python
def detect_outliers(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (column < lower_bound) | (column > upper_bound)

numeric_cols = df.select_dtypes(include=[np.number]).columns.drop(binary_columns)

for col in numeric_cols:
    print(f"{col}: {df[col].nunique()} unique values")
    outliers = detect_outliers(df[col])
    print('outliers: ', outliers.sum(), '\n')
```

```python
mask = detect_outliers(df['Year_Birth'])
display(df[mask])
df = df[~mask]
```

```python
mask = detect_outliers(df['Income'])
display(df[mask])
display(df['Income'].min()) #not included in outliers as you see
```

```python
Mnt = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(Mnt):
    mask = detect_outliers(df[col])
    
    axes[i].scatter(range(len(df)), df[col], color='blue', alpha=0.5, label='Normal')
    axes[i].scatter(df.index[mask], df.loc[mask, col], color='red', label='Outliers')
    
    axes[i].set_title(f'Outliers in {col}')
    axes[i].set_xlabel('Index')
    axes[i].set_ylabel(col)

handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper right')

plt.tight_layout()
plt.show()

# Mnt = ['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']

# for col in Mnt:
#     mask = detect_outliers(df[col])
#     display(df[mask])
```

```python
from datetime import datetime
df['Age'] = datetime.now().year - df['Year_Birth']
df['Age']
```

```python
df['Family_Size'] = df['Kidhome'] + df['Teenhome']
df['Family_Size']
```

```python
df['Total_Spending'] = df['MntWines'] + df['MntFruits'] + df['MntMeatProducts'] + df['MntFishProducts'] + df['MntSweetProducts'] + df['MntGoldProds']
df['Total_Spending']
```

```python
df['Wine_Ratio'] = df['MntWines'] / df['Total_Spending']
df['Gold_Ratio'] = df['MntGoldProds'] / df['Total_Spending']
```

```python
df['Total_Purchases'] = df['NumWebPurchases'] + df['NumCatalogPurchases'] + df['NumStorePurchases']
df['Deals_Ratio'] = df['NumDealsPurchases'] / df['Total_Purchases']
```

```python
df['Visit_to_Purchase_Ratio'] = df['NumWebVisitsMonth'] / df['Total_Purchases']
```

```python
df['Total_Accepted_Campaigns'] = df['AcceptedCmp1'] + df['AcceptedCmp2']  + df['AcceptedCmp3']  + df['AcceptedCmp4']  + df['AcceptedCmp5'] 
df['Is_Repeat_Responder'] = df['Total_Accepted_Campaigns'].apply(lambda x: 1 if x > 1 else 0)
```
